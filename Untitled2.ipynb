### 1. 가상 데이터 생성 (Synthetic Data Generation)
실제 위성 데이터(VIIRS, MODIS)의 물리적 특성을 모사한 부산 지역 1km 격자 수준의 가상 데이터를 생성하여 분석 환경을 구축합니다.

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

# 데이터 재현을 위한 시드 설정
np.random.seed(42)

# 1. 부산 지역 약 1,000개 격자 x 48개월(4년) 데이터 가상 생성
n_grids = 1000
years = [2020, 2021, 2022, 2023]
months = list(range(1, 13))

# 격자 고유 특성 정의
grid_ids = [f"grid_{i:04d}" for i in range(n_grids)]
elevation = np.random.exponential(scale=150, size=n_grids) # 고도 (평지~산악)
dist_to_coast = np.random.uniform(0.1, 20.0, size=n_grids) # 해안선 거리 (km)

data_list = []

for year in years:
    for month in months:
        # 계절 정의
        if month in [3, 4, 5]:
            season = 'Spring'
            base_temp = 12.0
        elif month in [6, 7, 8]:
            season = 'Summer'
            base_temp = 24.0
        elif month in [9, 10, 11]:
            season = 'Autumn'
            base_temp = 16.0
        else:
            season = 'Winter'
            base_temp = 2.0

        for i, grid_id in enumerate(grid_ids):
            # 야간조명 강도 (도심 중심부일수록 높고, 고도가 높을수록 낮음)
            elev_factor = np.exp(-elevation[i] / 100.0)
            coast_factor = np.exp(-dist_to_coast[i] / 10.0)
            ntl = np.random.gamma(shape=2, scale=10) * (0.3 + 0.4 * elev_factor + 0.3 * coast_factor)
            ntl = np.clip(ntl, 0.1, 100.0)

            # NDVI (계절 영향 및 조명/개발 지역일수록 낮음)
            ndvi_base = 0.6 if season in ['Spring', 'Summer'] else 0.4
            ndvi = ndvi_base - (ntl / 300.0) - (elevation[i] / 5000.0) + np.random.normal(0, 0.05)
            ndvi = np.clip(ndvi, 0.05, 0.9)

            # 지표면 온도 (LST): 기본온도 + 야간조명 효과(UHI) - 고도 효과 - 해안거리 효과 + 노이즈
            temp_effect = base_temp + (0.08 * ntl) - (0.006 * elevation[i]) - (0.1 * dist_to_coast[i]) + (0.5 * ndvi)
            lst = temp_effect + np.random.normal(0, 1.2)

            # 유효 관측 수 (기상 조건에 따른 월별 관측 성공 횟수: 0 ~ 30)
            valid_obs = np.random.randint(5, 31)

            data_list.append({
                'grid_id': grid_id,
                'year': year,
                'month': month,
                'season': season,
                'elevation': elevation[i],
                'dist_to_coast': dist_to_coast[i],
                'ntl': ntl,
                'ndvi': ndvi,
                'lst': lst,
                'valid_obs': valid_obs
            })

df = pd.DataFrame(data_list)
display(df.head())
print(f"Total records generated: {len(df)}")
```

### 2. 세부목적 1: 야간조명 강도와 야간 지표면온도의 단순 공간 연관성 정량화
환경 요소를 보정하기 전에, 야간조명 강도($NTL$)가 지표면온도($LST$)에 미치는 1차적인 연관성을 단변량 선형 회귀 분석을 통해 산출합니다.

```python
# 단순 회귀 분석 실행 (보정 전 단순 연관성)
y = df['lst']
X_simple = sm.add_constant(df['ntl'])

model_simple = sm.OLS(y, X_simple).fit()
print(model_simple.summary())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# [그래프 1] 세부목적 1: 단순 연관성 산점도 및 회귀선
plt.figure(figsize=(8, 5))
sns.regplot(
    data=df.sample(4000, random_state=42),  # 데이터 증가로 샘플링 수도 4,000개로 조정하여 가시성 확보
    x='ntl',
    y='lst',
    scatter_kws={'alpha': 0.15, 'color': 'gray'},
    line_kws={'color': '#d95f02', 'linewidth': 2.5}
)
plt.title('Step 1 Visualization: Simple Relationship (LST vs. NTL) - 4 Years Data', fontsize=12, fontweight='bold')
plt.xlabel('Nightlight Intensity (VIIRS NTL)')
plt.ylabel('Nighttime LST (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### 3. 세부목적 2: 환경조건 보정 전후 연관성 비교 및 야간조명의 추가 설명력 평가
해안까지의 거리, 고도, NDVI를 환경조건 변수로 제어한 다중회귀 분석을 수행하고, 야간조명이 들어갔을 때의 결정계수 변화량($\Delta R^2$)을 계산합니다.

```python
# 1. 환경 조건만 고려한 모형 (Baseline Model)
features_env = ['elevation', 'dist_to_coast', 'ndvi']
X_env = sm.add_constant(df[features_env])
model_env = sm.OLS(y, X_env).fit()

# 2. 환경 조건 + 야간조명 강도 모형 (Full Model)
features_full = ['elevation', 'dist_to_coast', 'ndvi', 'ntl']
X_full = sm.add_constant(df[features_full])
model_full = sm.OLS(y, X_full).fit()

# 결과 산출 및 비교
r2_env = model_env.rsquared
r2_full = model_full.rsquared
delta_r2 = r2_full - r2_env

print("=== 세부목적 2 분석 결과 ===")
print(f"환경 조건 모형 R²: {r2_env:.4f}")
print(f"전체 변수 모형 R²: {r2_full:.4f}")
print(f"야간조명 강도의 추가 설명력 (ΔR²): {delta_r2:.4f}")
print("\n--- 야간조명 보정 전후 회귀 계수 비교 ---")
print(f"보정 전 야간조명 계수 (Simple): {model_simple.params['ntl']:.4f}")
print(f"보정 후 야간조명 계수 (Full): {model_full.params['ntl']:.4f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# [그래프 2] 세부목적 2: 환경조건 통제 전후 효과 크기 변화 및 설명력 비교
plt.figure(figsize=(8, 5))
coef_data = pd.DataFrame({
    'Model': ['Before Control (Simple)', 'After Control (Full)'],
    'NTL Coefficient (β)': [model_simple.params['ntl'], model_full.params['ntl']]
})
sns.barplot(
    data=coef_data,
    x='Model',
    y='NTL Coefficient (β)',
    hue='Model',
    legend=False,
    palette='YlGnBu_r'
)
plt.title(f'Step 2 Visualization: Coefficient Shift (ΔR² = {delta_r2:.4f}) - 4 Years Data', fontsize=12, fontweight='bold')
plt.ylabel('Effect Size (β coefficient)')
plt.xlabel('')
for p in plt.gca().patches:
    plt.gca().annotate(f"{p.get_height():.4f}",
                    (p.get_x() + p.get_width() / 2., p.get_height() - 0.015),
                    ha='center', va='center', color='white', fontweight='bold',
                    xytext=(0, 5), textcoords='offset points')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### 4. 세부목적 3: 계절별 연관성 분석 및 최소 유효 관측 수 기준에 대한 민감도 분석
월별 데이터 수집의 불확실성을 고려하여, 최소 유효 관측 수 기준(`min_obs`)을 다르게 설정했을 때 계절별 야간조명 회귀 계수의 추정값이 어떻게 변하는지 확인합니다.

```python
# 유효 관측 수 임계값 리스트 설정
min_obs_thresholds = [5, 15, 25]
seasons = ['Spring', 'Summer', 'Autumn', 'Winter']

sensitivity_results = []

for limit in min_obs_thresholds:
    # 필터링 적용
    filtered_df = df[df['valid_obs'] >= limit]

    for season in seasons:
        season_data = filtered_df[filtered_df['season'] == season]

        # 샘플 수가 부족할 경우 예외 처리
        if len(season_data) > 10:
            X_season = sm.add_constant(season_data[features_full])
            y_season = season_data['lst']
            model_season = sm.OLS(y_season, X_season).fit()

            coef_ntl = model_season.params['ntl']
            p_val = model_season.pvalues['ntl']
            ci_lower, ci_upper = model_season.conf_int().loc['ntl']

            sensitivity_results.append({
                'Min_Valid_Obs': limit,
                'Season': season,
                'Sample_Size': len(season_data),
                'NTL_Coef': coef_ntl,
                'P_Value': p_val,
                'CI_Lower': ci_lower,
                'CI_Upper': ci_upper
            })

sensitivity_df = pd.DataFrame(sensitivity_results)
display(sensitivity_df)
```

### 5. 결과 시각화
민감도 분석 결과를 차트로 시각화하여 계절별 계수 차이와 필터링 강도에 따른 신뢰구간 변화를 직관적으로 비교합니다.

```python
plt.figure(figsize=(12, 6))
sns.pointplot(
    data=sensitivity_df,
    x='Season',
    y='NTL_Coef',
    hue='Min_Valid_Obs',
    dodge=0.3,
    linestyle='none',
    capsize=0.1
)

# 신뢰구간 에러바 수동 추가
for i, row in sensitivity_df.iterrows():
    # x좌표 매핑을 위한 계산
    season_idx = seasons.index(row['Season'])
    hue_idx = min_obs_thresholds.index(row['Min_Valid_Obs'])
    offset = (hue_idx - 1) * 0.1  # dodge 효과 수동 적용

    plt.errorbar(
        x=season_idx + offset,
        y=row['NTL_Coef'],
        yerr=[[row['NTL_Coef'] - row['CI_Lower']], [row['CI_Upper'] - row['NTL_Coef']]],
        fmt='none',
        color='gray',
        alpha=0.7
    )

plt.title('Seasonal Sensitivity Analysis of Nightlight Coefficients by Min Valid Observations - 4 Years Data')
plt.ylabel('Coefficient of Nightlight (NTL)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
